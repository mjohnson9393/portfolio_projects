```python
!pip install ucimlrepo
import pandas as pd
from ucimlrepo import fetch_ucirepo

autism = fetch_ucirepo(id=426)
df = pd.concat([autism.data.features, autism.data.targets], axis=1)
```

```python
df['result'].value_counts().sort_index()
```

```python
df['score_range'] = pd.cut(df['result'], bins=[-1, 3, 6, 10], labels=['0-3', '4-6', '7-10'])
(df['score_range'].value_counts(normalize=True).sort_index() * 100).round(1)
```

```python
df['gender'].value_counts()
```

```python
df_clean = df.dropna(subset=['age'])
df_clean = df_clean[df_clean['age'] <= 100]
df_clean.shape
```

```python
df_clean['gender'] = df_clean['gender'].map({'m': 'Male', 'f': 'Female'})
```

```python
df_clean['age_group'] = pd.cut(df_clean['age'], bins=[0, 29, 44, 100], labels=['17-29', '30-44', '45+'])
df_clean['age_group'].value_counts()
```

```python
df_clean.groupby('gender')['result'].mean().round(2)
```

```python
df_clean.groupby('age_group', observed=True)['result'].mean().round(2)
```

```python
df_clean.groupby(['age_group', 'gender'], observed=True)['result'].mean().round(2)
```

```python
df_clean['score_range'] = pd.cut(df_clean['result'], bins=[-1, 3, 6, 10], labels=['0-3', '4-6', '7-10'])
df_clean.to_csv('autism_data.csv', index=False)
from google.colab import files
files.download('autism_data.csv')
```
